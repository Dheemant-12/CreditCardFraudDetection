###How to create a Basic Array

```python
np.zeros(5) # used fror creating an array filled with zeros
np.ones(3) # similarly for ones also
np.empty(8) # this fills the array with most random numbers which doesnt usually change
np.arange(5) # used to make an array from range
np.arange(2,10,2) # syntax a(start,stop,step) , always start is included but end number is not included in output
np.linspace(0,10,num=5) #You can also use np.linspace() to create an array with values that are spaced linearly in a specified interval
np.ones(2,dtype=np.int64) # we can use np.float64 also
```

###Adding, removing, and sorting elements and concatenating


```python
arr=np.array([10,3,45,57,8])
np.sort(arr)#for ascending order
np.argsort(arr) # array([1, 4, 0, 2, 3])
#lexsort prints alphabetically and returns indices
first_names = ['Charlie', 'Alice', 'Bob']
surnames = ['Smith', 'Smith', 'Jones']
indices = np.lexsort((first_names, surnames))
print(indices)  # Output: [2, 1, 0]
#searchsorted
sorted_arr = np.array([10, 20, 30, 40])
index = np.searchsorted(sorted_arr, 25)
print(index)  # Output: 2
#partial
arr = np.array([90, 10, 50, 20, 80])
partitioned = np.partition(arr, 2)
print(partitioned)  # Output: [20, 10, 50, 90, 80]
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
np.concatenate((a,b),axis=0)#concatenation which means combining 2 arrays together
```
